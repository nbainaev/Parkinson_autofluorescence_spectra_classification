```python
import numpy as np
def standart_scale(data: pd.DataFrame | np.ndarray) -> pd.DataFrame | np.ndarray:

    if isinstance(data, pd.DataFrame):
        return data.sub(data.mean(axis=1), axis=0).div(data.std(axis=1), axis=0)
    else:
        mean = data.mean(axis=1, keepdims=True)
        std = data.std(axis=1, keepdims=True)

        return (data - mean) / std
```

```python
from src.preprocessing import i_norm
from sklearn.linear_model import LinearRegression, RANSACRegressor

def auc_normalization(spectra, dist):
    """
    Нормировка по площади под кривой - сохраняет форму и отношения
    """
    areas = np.trapezoid(spectra, axis=1)  # площадь под каждым спектром
    mean_area = np.mean(areas)
    normalization_factors = areas ** 2 /mean_area
    
    normalized = spectra * dist / normalization_factors[:, np.newaxis]
    return normalized

def i_norm(data: pd.DataFrame, rename: bool=True) -> pd.DataFrame:
    data_Inorm_arr = []
    for group in data.index.get_level_values("GROUP").unique():
        data_buf = data[data.index.get_level_values("GROUP").isin([group])]
        mean_values = data_buf.mean().to_numpy().reshape(-1, 1)
        x = auc_normalization(data_buf.to_numpy(), dist=data_buf.columns[-1] - data_buf.columns[0])
        data_Inorm_arr.extend(x)

    data_Inorm = np.asarray(data_Inorm_arr)
    data_Inorm = pd.DataFrame(data_Inorm, columns=data.keys(), index=data.index)
    if rename:
        data_Inorm.columns = [f"I{i[1:]}" if str(i)[0] == "D" else f"I{i}" for i in data_Inorm.keys()]
    return data_Inorm
df_inorm = i_norm(df[df.columns[:150]], rename=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from typing import Optional, List, Union

def plot_age_distribution_by_sex(df: pd.DataFrame, 
                                age_column: str = 'AGE',
                                sex_column: str = 'SEX',
                                group_column: Optional[str] = None,
                                groups: Union[str, List[str]] = 'all',
                                bins: int = 20,
                                figsize: tuple = (12, 6),
                                palette: str = 'viridis',
                                alpha: float = 0.7,
                                kde: bool = True,
                                title: str = None):
    """
    Строит гистограмму распределения по возрастам и полу с опцией группировки по экспериментальным группам.
    
    Parameters:
    -----------
    df : pd.DataFrame
        DataFrame с данными
    age_column : str
        Название колонки с возрастом (по умолчанию 'age')
    sex_column : str
        Название колонки с полом (по умолчанию 'sex')
    group_column : str, optional
        Название колонки с экспериментальными группами. Если None, группировка не применяется.
    groups : str or list
        Какие группы отображать: 'all' - все группы, либо список конкретных групп.
    bins : int
        Количество бинов для гистограммы
    figsize : tuple
        Размер фигуры
    palette : str
        Цветовая палитра seaborn
    alpha : float
        Прозрачность гистограмм
    kde : bool
        Добавлять ли KDE (ядровую оценку плотности)
    title : str, optional
        Заголовок графика
    """
    
    # Создаем копию данных чтобы не менять оригинал
    data = df.copy()
    
    # Проверяем наличие необходимых колонок
    required_columns = [age_column, sex_column]
    if group_column:
        required_columns.append(group_column)
    
    missing_columns = [col for col in required_columns if col not in data.columns]
    if missing_columns:
        raise ValueError(f"Отсутствуют колонки: {missing_columns}")
    
    # Фильтруем данные по группам если нужно
    if group_column and groups != 'all':
        if isinstance(groups, str):
            groups = [groups]
        data = data[data[group_column].isin(groups)]
    
    # Настраиваем стиль
    plt.style.use('seaborn-v0_8')
    sns.set_palette(palette)
    
    # Создаем фигуру
    fig, axes = plt.subplots(1, 2, figsize=figsize, sharey=True)
    
    if group_column:
        # Случай с группировкой по экспериментальным группам
        unique_groups = data[group_column].unique()
        n_groups = len(unique_groups)
        
        for i, group in enumerate(unique_groups):
            group_data = data[data[group_column] == group]
            
            # Распределение для мужчин
            male_data = group_data[group_data[sex_column] == 'Male'][age_column]
            if len(male_data) > 0:
                sns.histplot(data=male_data, ax=axes[0], label=f'{group} (M)', 
                           bins=bins, alpha=alpha, kde=kde)
            
            # Распределение для женщин
            female_data = group_data[group_data[sex_column] == 'Female'][age_column]
            if len(female_data) > 0:
                sns.histplot(data=female_data, ax=axes[1], label=f'{group} (F)', 
                           bins=bins, alpha=alpha, kde=kde)
                
    else:
        # Случай без группировки - все данные вместе
        male_data = data[data[sex_column] == 'Male'][age_column]
        female_data = data[data[sex_column] == 'Female'][age_column]
        
        if len(male_data) > 0:
            sns.histplot(data=male_data, ax=axes[0], label='Мужчины', 
                       bins=bins, alpha=alpha, kde=kde, color='skyblue')
        
        if len(female_data) > 0:
            sns.histplot(data=female_data, ax=axes[1], label='Женщины', 
                       bins=bins, alpha=alpha, kde=kde, color='lightcoral')
    
    # Настраиваем внешний вид
    axes[0].set_title('Распределение по возрастам (Мужчины)')
    axes[0].set_xlabel('Возраст')
    axes[0].set_ylabel('Количество')
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)
    
    axes[1].set_title('Распределение по возрастам (Женщины)')
    axes[1].set_xlabel('Возраст')
    axes[1].set_ylabel('')
    axes[1].legend()
    axes[1].grid(True, alpha=0.3)
    
    # Общий заголовок
    if title:
        plt.suptitle(title, fontsize=16, fontweight='bold')
    else:
        group_info = f" - Группы: {', '.join(unique_groups)}" if group_column else ""
        plt.suptitle(f'Распределение возраста по полу{group_info}', 
                    fontsize=16, fontweight='bold')
    
    plt.tight_layout()
    plt.show()
    
    # Выводим статистику
    print("Статистика по данным:")
    print(f"Общее количество наблюдений: {len(data)}")
    print(f"Мужчины: {len(data[data[sex_column] == 'Male'])}")
    print(f"Женщины: {len(data[data[sex_column] == 'Female'])}")
    
    if group_column:
        print("\nПо группам:")
        for group in data[group_column].unique():
            group_data = data[data[group_column] == group]
            print(f"{group}: всего {len(group_data)}, "
                  f"М: {len(group_data[group_data[sex_column] == 'Male'])}, "
                  f"Ж: {len(group_data[group_data[sex_column] == 'Female'])}")

# Альтернативная версия с одним графиком и facet grid
def plot_age_distribution_facet(df: pd.DataFrame,
                               age_column: str = 'age',
                               sex_column: str = 'sex', 
                               group_column: Optional[str] = None,
                               bins: int = 15,
                               height: int = 5,
                               aspect: float = 1.5):
    """
    Версия с facet grid для более компактного отображения.
    """
    
    data = df.copy()
    
    if group_column:
        # Используем facet grid для группировки
        g = sns.FacetGrid(data, col=group_column, hue=sex_column, 
                         height=height, aspect=aspect, sharex=True, sharey=True)
        g.map(sns.histplot, age_column, bins=bins, alpha=0.7, kde=True)
        g.add_legend()
        g.set_axis_labels('Возраст', 'Количество')
        
    else:
        # Простой график без группировки
        plt.figure(figsize=(10, 6))
        sns.histplot(data=data, x=age_column, hue=sex_column, 
                    bins=bins, alpha=0.7, kde=True)
        plt.title('Распределение возраста по полу')
        plt.xlabel('Возраст')
        plt.ylabel('Количество')
        plt.legend(title='Пол')
        plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

print("=== Простое распределение по полу ===")
plot_age_distribution_by_sex(sa_df, title="Общее распределение по возрастам")
print("\n=== С группировкой по всем группам ===")
plot_age_distribution_by_sex(sa_df, 
                            group_column='GROUP',
                            groups='all',
                            palette='husl',
                            title="Распределение по возрастам и экспериментальным группам")

# plot_age_distribution_by_sex(sa_df,
#                             group_column='group', 
#                             groups=['Control', 'Treatment_A'],
#                             title="Распределение для Control и Treatment_A")
# Пример использования
# if __name__ == "__main__":
#     # Создаем пример данных
#     np.random.seed(42)
#     n = 1000
    
#     example_data = pd.DataFrame({
#         'age': np.random.normal(45, 15, n),
#         'sex': np.random.choice(['M', 'F'], n, p=[0.5, 0.5]),
#         'group': np.random.choice(['Control', 'Treatment_A', 'Treatment_B'], n)
#     })
    
#     # Пример 1: Простое распределение по полу
#     print("=== Простое распределение по полу ===")
#     plot_age_distribution_by_sex(example_data, title="Общее распределение по возрастам")
    
#     # Пример 2: С группировкой по всем экспериментальным группам
#     print("\n=== С группировкой по всем группам ===")
#     plot_age_distribution_by_sex(example_data, 
#                                 group_column='group',
#                                 groups='all',
#                                 title="Распределение по возрастам и экспериментальным группам")
    
#     # Пример 3: Только для конкретных групп
#     print("\n=== Только для Control и Treatment_A ===")
#     plot_age_distribution_by_sex(example_data,
#                                 group_column='group', 
#                                 groups=['Control', 'Treatment_A'],
#                                 title="Распределение для Control и Treatment_A")
    
#     # Пример 4: Facet grid версия
#     print("\n=== Facet grid версия ===")
#     plot_age_distribution_facet(example_data, group_column='group')
```

```python
import scipy.stats as stats
def compute_t_Welch(data, start, stop, alpha=0.05):
    """
        Функция для рассчета t-оценок по подгруппам Паркинсон, Контроль, Сравнение
        
        data: pd.DataFrame - данные
        start: pd.Index - начало исследуемого участка спектра
        stop: pd.Index - конец исследуемого участка
    """
    t_cont_park = {"t_value": [], "p_value": [], "df": [], 'crit_t': []}
    t_comp_park = {"t_value": [], "p_value": [], "df": [], 'crit_t': []}
    t_comp_cont = {"t_value": [], "p_value": [], "df": [], 'crit_t': []}
    df_cont = data.loc[data.index.get_level_values(level='GROUP') == "Control", start:stop].copy()
    df_park = data.loc[data.index.get_level_values(level='GROUP') == "Parkinson", start:stop].copy()
    df_comp = data.loc[data.index.get_level_values(level='GROUP') == "Comparison", start:stop].copy()
    try:
        preds = [int(val)for val in df_cont.keys()]
    except:
        preds = [int(val[1:])for val in df_cont.keys()]
    
    for predictor in df_cont.keys():
        res = stats.ttest_ind(df_cont[predictor], df_park[predictor], equal_var=False, alternative="two-sided")
        t, p, df = res[0], res[1], res.df
        critical_value_two_sided = stats.t.ppf(1 - alpha/2, df)
        t_cont_park['t_value'].append(t)
        t_cont_park['p_value'].append(p)
        t_cont_park['df'].append(df)
        t_cont_park['crit_t'].append(critical_value_two_sided)

    for predictor in df_cont.keys():
        res = stats.ttest_ind(df_comp[predictor], df_park[predictor], equal_var=False, alternative="two-sided")
        t, p, df = res[0], res[1], res.df
        critical_value_two_sided = stats.t.ppf(1 - alpha/2, df)
        t_comp_park['t_value'].append(t)
        t_comp_park['p_value'].append(p)
        t_comp_park['df'].append(df)
        t_comp_park['crit_t'].append(critical_value_two_sided)

    for predictor in df_cont.keys():
        res = stats.ttest_ind(df_comp[predictor], df_cont[predictor], equal_var=False, alternative="two-sided")
        t, p, df = res[0], res[1], res.df
        critical_value_two_sided = stats.t.ppf(1 - alpha/2, df)
        t_comp_cont['t_value'].append(t)
        t_comp_cont['p_value'].append(p)
        t_comp_cont['df'].append(df)
        t_comp_cont['crit_t'].append(critical_value_two_sided)
    
    return {'cont_park': t_cont_park, 'comp_park': t_comp_park, 'comp_cont': t_comp_cont}, preds
```

```python
def compute_Z(data, start, stop):
    """
        Функция для рассчета Z-оценок по подгруппам Паркинсон, Контроль, Сравнение
        
        data: pd.DataFrame - данные
        start: pd.Index - начало исследуемого участка спектра
        stop: pd.Index - конец исследуемого участка
    """
    Z_cont_park = []
    Z_comp_park = []
    Z_comp_cont = []
    df_cont = data.loc[data.index.get_level_values(level='GROUP') == "Control", start:stop].copy()
    df_park = data.loc[data.index.get_level_values(level='GROUP') == "Parkinson", start:stop].copy()
    df_comp = data.loc[data.index.get_level_values(level='GROUP') == "Comparison", start:stop].copy()
    try:
        preds = [int(val)for val in df_cont.keys()]
    except:
        preds = [int(val[1:])for val in df_cont.keys()]
    
    n1 = len(df_cont)
    n2 = len(df_park)
    mU = n1 * n2 / 2
    sigma_U = (n1 * n2 * (n1 + n2 + 1) / 12) ** 0.5
    for predictor in df_cont.keys():
        U, p = stats.mannwhitneyu(df_cont[predictor], df_park[predictor])
        Z_cont_park.append((float(U) - mU) / sigma_U)
    n1 = len(df_comp)
    n2 = len(df_park)
    mU = n1 * n2 / 2
    sigma_U = (n1 * n2 * (n1 + n2 + 1) / 12) ** 0.5
    for predictor in df_cont.keys():
        U, p = stats.mannwhitneyu(df_comp[predictor], df_park[predictor])
        Z_comp_park.append((float(U) - mU) / sigma_U)
    n1 = len(df_comp)
    n2 = len(df_cont)
    mU = n1 * n2 / 2
    sigma_U = (n1 * n2 * (n1 + n2 + 1) / 12) ** 0.5
    for predictor in df_cont.keys():
        U, p = stats.mannwhitneyu(df_comp[predictor], df_cont[predictor])
        Z_comp_cont.append((float(U) - mU) / sigma_U)
    
    return Z_cont_park, Z_comp_park, Z_comp_cont, preds
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import signal
from scipy.integrate import simpson as simps

def extract_spectral_features(spectra, frequencies):
    """
    Извлечение характеристических признаков из спектров
    """
    features = []
    
    for spec in spectra:
        feature_dict = {
            'max_intensity': np.max(spec),
            'min_intensity': np.min(spec),
            'mean_intensity': np.mean(spec),
            'std_intensity': np.std(spec),
            'total_area': simps(spec, frequencies),  # площадь под кривой
            'peak_frequency': frequencies[np.argmax(spec)],  # частота пика
        }
        
        # Поиск пиков
        peaks, properties = signal.find_peaks(spec, height=np.mean(spec))
        feature_dict.update({
            'n_peaks': len(peaks),
            'peak_heights': properties['peak_heights'] if len(peaks) > 0 else [0],
            'peak_positions': frequencies[peaks] if len(peaks) > 0 else [0]
        })
        
        # Центроида спектра (центр масс)
        centroid = np.sum(frequencies * spec) / np.sum(spec)
        feature_dict['centroid'] = centroid
        
        features.append(feature_dict)
    
    return pd.DataFrame(features)

def spectral_analysis(spectra, sampling_rate=1.0):
    """
    Анализ спектров с помощью методов Фурье
    """
    results = []
    
    for i, spec in enumerate(spectra):
        # Фурье-преобразование самого спектра (для анализа формы)
        fft_result = np.fft.fft(spec)
        fft_freq = np.fft.fftfreq(len(spec), d=1/sampling_rate)
        
        # Основные частотные компоненты
        power_spectrum = np.abs(fft_result)**2
        dominant_freq = fft_freq[np.argmax(power_spectrum[1:]) + 1]
        
        results.append({
            'spectrum_id': i,
            'dominant_frequency': dominant_freq,
            'spectral_entropy': calculate_spectral_entropy(power_spectrum),
            'bandwidth': calculate_bandwidth(power_spectrum, fft_freq)
        })
    
    return pd.DataFrame(results)

def calculate_spectral_entropy(power_spectrum):
    """Расчет спектральной энтропии"""
    norm_spectrum = power_spectrum / np.sum(power_spectrum)
    return -np.sum(norm_spectrum * np.log(norm_spectrum + 1e-12))

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

def pca_spectral_analysis(spectra_matrix, n_components=2):
    """
    PCA для снижения размерности спектральных данных
    """
    # Стандартизация
    scaler = StandardScaler()
    spectra_scaled = scaler.fit_transform(spectra_matrix)
    
    # PCA
    pca = PCA(n_components=n_components)
    principal_components = pca.fit_transform(spectra_scaled)
    
    print(f"Объясненная дисперсия: {pca.explained_variance_ratio_}")
    
    return principal_components, pca, scaler

from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score

def cluster_spectra(spectra, n_clusters=3, method='kmeans'):
    """
    Кластеризация спектров для выявления паттернов
    """
    if method == 'kmeans':
        clusterer = KMeans(n_clusters=n_clusters, random_state=42)
    elif method == 'dbscan':
        clusterer = DBSCAN(eps=0.5, min_samples=5)
    
    labels = clusterer.fit_predict(spectra)
    
    # Оценка качества кластеризации
    if len(set(labels)) > 1:
        score = silhouette_score(spectra, labels)
        print(f"Silhouette Score: {score:.3f}")
    
    return labels

from scipy.spatial.distance import cdist
from sklearn.metrics.pairwise import cosine_similarity

def spectral_similarity_analysis(spectra1, spectra2):
    """
    Анализ сходства между наборами спектров
    """
    # Евклидово расстояние
    euclidean_dist = cdist(spectra1, spectra2, metric='euclidean')
    
    # Косинусное сходство
    cosine_sim = cosine_similarity(spectra1, spectra2)
    
    # Корреляция Пирсона
    correlation_matrix = np.corrcoef(spectra1, spectra2)
    
    return {
        'euclidean_distance': euclidean_dist,
        'cosine_similarity': cosine_sim,
        'correlation': correlation_matrix
    }

def comprehensive_spectral_analysis(spectra, frequencies, labels=None):
    """
    Комплексный анализ спектральных данных
    """
    # 1. Извлечение признаков
    features_df = extract_spectral_features(spectra, frequencies)
    
    # 2. PCA для визуализации
    principal_components, pca, scaler = pca_spectral_analysis(spectra)
    
    # 3. Кластеризация
    cluster_labels = cluster_spectra(spectra, n_clusters=3)
    
    # 4. Визуализация
    plt.figure(figsize=(15, 10))
    
    # Исходные спектры
    plt.subplot(2, 3, 1)
    for i, spec in enumerate(spectra[:10]):  # первые 10 спектров
        plt.plot(frequencies, spec, alpha=0.7)
    plt.title('Исходные спектры')
    plt.xlabel('Частота/Длина волны')
    plt.ylabel('Интенсивность')
    
    # PCA проекция
    plt.subplot(2, 3, 2)
    scatter = plt.scatter(principal_components[:, 0], principal_components[:, 1], 
                         c=cluster_labels, cmap='viridis', alpha=0.7)
    plt.colorbar(scatter)
    plt.title('PCA проекция спектров')
    plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.2%})')
    plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.2%})')
    
    # Гистограмма признаков
    plt.subplot(2, 3, 3)
    plt.hist(features_df['centroid'], bins=20, alpha=0.7)
    plt.title('Распределение центроидов')
    plt.xlabel('Центроид')
    plt.ylabel('Частота')
    
    # Heatmap корреляций
    plt.subplot(2, 3, 4)
    correlation_matrix = np.corrcoef(spectra)
    plt.imshow(correlation_matrix, cmap='coolwarm', aspect='auto')
    plt.colorbar()
    plt.title('Матрица корреляций спектров')
    
    # Средние спектры по кластерам
    plt.subplot(2, 3, 5)
    for cluster_id in np.unique(cluster_labels):
        cluster_spectra_ = spectra[cluster_labels == cluster_id]
        mean_spectrum = np.mean(cluster_spectra_, axis=0)
        plt.plot(frequencies, mean_spectrum, label=f'Cluster {cluster_id}', linewidth=2)
    plt.title('Средние спектры по кластерам')
    plt.legend()
    
    plt.tight_layout()
    plt.show()
    
    return {
        'features': features_df,
        'pca_components': principal_components,
        'cluster_labels': cluster_labels,
        'pca_model': pca
    }
```

```python
from sklearn.model_selection import train_test_split

def train_test_val_split(spectra: pd.DataFrame, split_col: str='PARKINSON', val_split: bool=True, val_size: float=0.2, test_size: float=0.2, seed: int=42) -> tuple:
    idx = pd.IndexSlice
    y = pd.Series([1 if x == "YES" else 0 for x in spectra[split_col].to_list()], index = spectra.index)
    X = spectra.drop(columns=split_col).copy()
    y_buf = y.groupby(level='MG ID').max()
    if val_split:
        y_train, y_temp = train_test_split(y_buf, stratify=y_buf, test_size=test_size + val_size, random_state=seed)
        y_val, y_test = train_test_split(y_temp, stratify=y_temp, test_size=test_size / (val_size + test_size), random_state=seed)
        X_val = X.loc[idx[y_val.index, :, :]]
        y_val = y.loc[idx[y_val.index, :, :]]
    else:
        y_train, y_test = train_test_split(y_buf, stratify=y_buf, test_size=test_size, random_state=seed)
    X_train = X.loc[idx[y_train.index, :, :]]
    y_train = y.loc[idx[y_train.index, :, :]]
    X_test = X.loc[idx[y_test.index, :, :]]
    y_test = y.loc[idx[y_test.index, :, :]]
    
    if val_split:
        return X_train, y_train, X_val, y_val, X_test, y_test
    else:
        return X_train, y_train, X_test, y_test
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score, roc_curve
from matplotlib import rcParams

def draw_roc_curve(y_true: np.ndarray, 
                   proba_all: list, 
                   y_pred_all: list, 
                   names: list,
                   figsize=(10, 8),
                   linewidth=2,
                   diagonal_alpha=0.5,
                   style='ggplot',
                   palette='viridis'):
    """
    Улучшенная визуализация ROC-кривых для нескольких моделей
    
    Параметры:
    ----------
    y_true : np.ndarray
        Истинные метки классов (1D массив)
    proba_all : list
        Список предсказанных вероятностей для положительного класса
    y_pred_all : list
        Список предсказанных классов
    names : list
        Названия моделей (для легенды)
    figsize : tuple
        Размер графика (ширина, высота)
    linewidth : int
        Толщина линий ROC-кривых
    diagonal_alpha : float
        Прозрачность диагональной пунктирной линии
    style : str
        Стиль matplotlib (доступные: 'ggplot', 'seaborn', 'default' и др.)
    palette : str
        Название цветовой палитры ('viridis', 'plasma', 'tab10' и др.)
    """
    
    # Проверка доступных стилей
    available_styles = plt.style.available
    if style not in available_styles:
        print(f"Стиль '{style}' недоступен. Используются стандартные настройки.")
        print(f"Доступные стили: {available_styles}")
        style = 'default'
    
    # Применение стиля
    plt.style.use(style)
    
    # Настройка параметров
    rcParams.update({
        'axes.titlesize': 20,
        'axes.labelsize': 16,
        'xtick.labelsize': 12,
        'ytick.labelsize': 12,
        'legend.fontsize': 14,
        'grid.alpha': 0.3
    })
    
    # Создание фигуры
    fig, ax = plt.subplots(figsize=figsize)
    
    # Диагональная линия (random classifier)
    ax.plot([0, 1], [0, 1], 
            linestyle='--', 
            color='gray', 
            alpha=diagonal_alpha,
            label='Случайный классификатор (AUC = 0.5)')
    
    # Получение цветов из палитры
    # colors = plt.get_cmap(palette, len(names))
    colors = [
    "#E6194B",  # Красный
    "#3CB44B",  # Зелёный
    "#4363D8",  # Синий
    "#FFE119",  # Жёлтый
    "#F032E6",  # Пурпурный
    "#42D4F4"   # Голубой
]
    
    # Построение ROC-кривых для каждой модели
    for i, (proba, y_pred, name) in enumerate(zip(proba_all, y_pred_all, names)):
        auc_score = roc_auc_score(y_true, proba)
        fpr, tpr, _ = roc_curve(y_true, proba, pos_label=1)
        
        ax.plot(fpr, tpr, 
                linewidth=linewidth,
                color=colors[i],
                label=f'{name} (AUC = {auc_score:.3f})')
    
    # Настройка отображения
    ax.set_title('ROC-кривые моделей', pad=20)
    ax.set_xlabel('False Positive Rate (FPR)')
    ax.set_ylabel('True Positive Rate (TPR)')
    ax.grid(True)
    ax.legend(frameon=True, framealpha=0.9, fontsize=16)
    
    # Добавление информации о данных
    # ax.annotate(f'Общее количество: {len(y_true)}\nПоложительный класс: {sum(y_true)}\nОтрицательный класс: {len(y_true)-sum(y_true)}',
    #             xy=(0.98, 0.02), 
    #             xycoords='axes fraction',
    #             ha='right', 
    #             va='bottom',
    #             fontsize=12,
    #             bbox=dict(boxstyle='round', alpha=0.2))
    
    # Установка границ
    ax.set_xlim([-0.01, 1.01])
    ax.set_ylim([-0.01, 1.01])
    
    # Улучшенное отображение
    plt.tight_layout()
    plt.show()
```

```python
from sklearn.metrics import precision_score, accuracy_score, recall_score, confusion_matrix
def print_metrics(y_true, y_pred, name="Undefined"):

    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    n_bootstraps = 1000
    metrics = {
        'accuracy': [],
        'precision': [],
        'recall': [],
        'specificity': []
    }

    for _ in range(n_bootstraps):
        indices = np.random.choice(len(y_true), len(y_true), replace=True)
        y_true_boot = y_true[indices]
        y_pred_boot = y_pred[indices]
        tn, fp, fn, tp = confusion_matrix(y_true_boot, y_pred_boot).ravel()
        metrics['accuracy'].append(accuracy_score(y_true_boot, y_pred_boot))
        metrics['precision'].append(precision_score(y_true_boot, y_pred_boot, zero_division=0))
        metrics['recall'].append(recall_score(y_true_boot, y_pred_boot, zero_division=0))
        metrics['specificity'].append(tn / (tn+fp))

    # Стандартное отклонение
    std_dev = {key: np.std(values) for key, values in metrics.items()}
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    print(f"{name} classifier:")
    print(f"Accuracy: {accuracy_score(y_true, y_pred):.2f} ± {std_dev['accuracy']:.2f}")
    print(f"Precision: {precision_score(y_true, y_pred):.2f} ± {std_dev['precision']:.2f}")
    print(f"Recall: {recall_score(y_true, y_pred):.2f} ± {std_dev['recall']:.2f}")
    print(f"Specificity: {tn / (tn+fp):.2f} ± {std_dev['specificity']:.2f}")
    print("-" * 50)
    print(f"Confusion matrix: ")
    print(f'{int(tp), int(fp)}')
    print(f'{int(fn), int(tn)}')
    print("-" * 50)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, 
    roc_auc_score, confusion_matrix, classification_report,
    precision_recall_curve, roc_curve
)
from sklearn.base import clone
import warnings
warnings.filterwarnings('ignore')

class CustomCrossValidator:
    """
    Кастомный класс для обучения с кросс-валидацией и подсчетом метрик
    """
    
    def __init__(self, model, n_splits=5, random_state=42, scoring_metrics=None):
        """
        Parameters:
        -----------
        model : object
            Модель с интерфейсом scikit-learn (fit, predict, predict_proba)
        n_splits : int
            Количество фолдов для кросс-валидации
        random_state : int
            Seed для воспроизводимости
        scoring_metrics : dict
            Словарь с дополнительными метриками {имя: функция}
        """
        self.model = model
        self.n_splits = n_splits
        self.random_state = random_state
        self.scoring_metrics = scoring_metrics or {}
        self.cv_results_ = {}
        self.models_ = []
        self.fold_predictions_ = []
        
        # Стандартные метрики
        self.base_metrics = {
            'accuracy': accuracy_score,
            'precision': precision_score,
            'recall': recall_score,
            'f1': f1_score,
            'roc_auc': roc_auc_score
        }
        
        # Комбинируем все метрики
        self.all_metrics = {**self.base_metrics, **self.scoring_metrics}
        
    def _calculate_specificity(self, y_true, y_pred):
        """Расчет специфичности (True Negative Rate)"""
        tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
        return tn / (tn + fp) if (tn + fp) > 0 else 0
    
    def _calculate_metrics(self, y_true, y_pred, y_proba=None):
        """Расчет всех метрик для одного фолда"""
        metrics = {}
        
        # Базовые метрики
        for name, metric_func in self.base_metrics.items():
            if name == 'roc_auc':
                if y_proba is not None and len(np.unique(y_true)) > 1:
                    metrics[name] = metric_func(y_true, y_proba)
                else:
                    metrics[name] = np.nan
            else:
                metrics[name] = metric_func(y_true, y_pred)
        
        # Специфичность
        metrics['specificity'] = self._calculate_specificity(y_true, y_pred)
        
        # Дополнительные метрики
        for name, metric_func in self.scoring_metrics.items():
            metrics[name] = metric_func(y_true, y_pred)
            
        return metrics
    
    def fit(self, X, y, verbose=True):
        """
        Обучение модели с кросс-валидацией
        
        Parameters:
        -----------
        X : array-like of shape (n_samples, n_features)
            Матрица признаков
        y : array-like of shape (n_samples,)
            Вектор целевой переменной
        verbose : bool
            Выводить ли прогресс
        """
        # Инициализация структур для хранения результатов
        self.cv_results_ = {metric: [] for metric in list(self.all_metrics.keys()) + ['specificity']}
        self.cv_results_['train_metrics'] = []
        self.models_ = []
        self.fold_predictions_ = []
        
        # Стратифицированная кросс-валидация
        skf = StratifiedKFold(n_splits=self.n_splits, shuffle=True, 
                             random_state=self.random_state)
        
        if verbose:
            print(f"🚀 Начало кросс-валидации ({self.n_splits} фолдов)")
            print("=" * 60)
        
        for fold, (train_idx, val_idx) in enumerate(skf.split(X, y)):
            if verbose:
                print(f"📊 Фолд {fold + 1}/{self.n_splits}")
            
            # Разделение данных
            X_train, X_val = X[train_idx], X[val_idx]
            y_train, y_val = y[train_idx], y[val_idx]
            
            # Клонирование и обучение модели
            model = clone(self.model)
            model.fit(X_train, y_train)
            self.models_.append(model)
            
            # Предсказания
            y_pred = model.predict(X_val)
            y_proba = model.predict_proba(X_val)[:, 1] if hasattr(model, 'predict_proba') else None
            
            # Сохраняем предсказания
            fold_results = {
                'fold': fold,
                'val_idx': val_idx,
                'y_true': y_val,
                'y_pred': y_pred,
                'y_proba': y_proba
            }
            self.fold_predictions_.append(fold_results)
            
            # Расчет метрик
            val_metrics = self._calculate_metrics(y_val, y_pred, y_proba)
            
            # Метрики на тренировочных данных (для анализа переобучения)
            y_train_pred = model.predict(X_train)
            y_train_proba = model.predict_proba(X_train)[:, 1] if hasattr(model, 'predict_proba') else None
            train_metrics = self._calculate_metrics(y_train, y_train_pred, y_train_proba)
            
            # Сохранение результатов
            for metric_name, metric_value in val_metrics.items():
                self.cv_results_[metric_name].append(metric_value)
            
            self.cv_results_['train_metrics'].append(train_metrics)
            
            if verbose:
                print(f"   ✅ Accuracy: {val_metrics['accuracy']:.3f}, "
                      f"Precision: {val_metrics['precision']:.3f}, "
                      f"Recall: {val_metrics['recall']:.3f}")
                print(f"   📈 ROC-AUC: {val_metrics.get('roc_auc', 'N/A'):.3f}, "
                      f"Specificity: {val_metrics['specificity']:.3f}")
        
        if verbose:
            print("=" * 60)
            self._print_summary()
    
    def _print_summary(self):
        """Печать сводки по кросс-валидации"""
        print("📊 СВОДКА ПО КРОСС-ВАЛИДАЦИИ")
        print("=" * 50)
        
        for metric in ['accuracy', 'precision', 'recall', 'specificity', 'f1', 'roc_auc']:
            if metric in self.cv_results_ and len(self.cv_results_[metric]) > 0:
                scores = self.cv_results_[metric]
                mean_score = np.mean(scores)
                std_score = np.std(scores)
                print(f"{metric.capitalize()}: {mean_score:.3f} ± {std_score:.3f}")
        
        # Анализ переобучения
        if self.cv_results_['train_metrics']:
            train_acc = np.mean([m['accuracy'] for m in self.cv_results_['train_metrics']])
            val_acc = np.mean(self.cv_results_['accuracy'])
            overfitting_gap = train_acc - val_acc
            print(f"📉 Разрыв train/val accuracy: {overfitting_gap:.3f}")
            
            if overfitting_gap > 0.05:
                print("⚠️  Возможно переобучение!")
            elif overfitting_gap < 0.01:
                print("✅ Модель хорошо обобщает")
    
    def get_summary(self):
        """Возвращает сводку метрик в виде DataFrame"""
        summary_data = {}
        for metric in ['accuracy', 'precision', 'recall', 'specificity', 'f1', 'roc_auc']:
            if metric in self.cv_results_:
                scores = self.cv_results_[metric]
                summary_data[metric] = {
                    'mean': np.mean(scores),
                    'std': np.std(scores),
                    'min': np.min(scores),
                    'max': np.max(scores)
                }
        
        return pd.DataFrame(summary_data).T
    
    def plot_metrics(self, figsize=(15, 10)):
        """Визуализация метрик по фолдам"""
        fig, axes = plt.subplots(2, 3, figsize=figsize)
        axes = axes.ravel()
        
        metrics_to_plot = ['accuracy', 'precision', 'recall', 'specificity', 'f1', 'roc_auc']
        
        for i, metric in enumerate(metrics_to_plot):
            if i >= len(axes):
                break
                
            if metric in self.cv_results_:
                # Метрики валидации
                val_scores = self.cv_results_[metric]
                
                # Метрики тренировки (если доступны)
                train_scores = [m.get(metric, np.nan) for m in self.cv_results_['train_metrics']]
                
                x_pos = np.arange(len(val_scores))
                width = 0.35
                
                axes[i].bar(x_pos - width/2, train_scores, width, label='Train', alpha=0.7)
                axes[i].bar(x_pos + width/2, val_scores, width, label='Validation', alpha=0.7)
                
                axes[i].set_title(f'{metric.capitalize()} по фолдам')
                axes[i].set_xlabel('Фолд')
                axes[i].set_ylabel(metric.capitalize())
                axes[i].set_xticks(x_pos)
                axes[i].set_xticklabels([f'Fold {i+1}' for i in range(len(val_scores))])
                axes[i].legend()
                axes[i].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
    
    def plot_confusion_matrix(self, normalize=True, figsize=(12, 8)):
        """Визуализация confusion matrix по фолдам"""
        n_cols = min(3, self.n_splits)
        n_rows = (self.n_splits + n_cols - 1) // n_cols
        
        fig, axes = plt.subplots(n_rows, n_cols, figsize=figsize)
        if n_rows == 1:
            axes = [axes] if n_cols == 1 else axes
        else:
            axes = axes.ravel()
        
        for i, fold_result in enumerate(self.fold_predictions_):
            if i >= len(axes):
                break
                
            y_true = fold_result['y_true']
            y_pred = fold_result['y_pred']
            
            cm = confusion_matrix(y_true, y_pred)
            if normalize:
                cm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
                fmt = '.2f'
            else:
                fmt = 'd'
            
            sns.heatmap(cm, annot=True, fmt=fmt, cmap='Blues', ax=axes[i],
                       xticklabels=['Pred 0', 'Pred 1'], 
                       yticklabels=['True 0', 'True 1'])
            axes[i].set_title(f'Fold {i+1}')
        
        # Скрываем лишние subplots
        for i in range(len(self.fold_predictions_), len(axes)):
            axes[i].set_visible(False)
        
        plt.tight_layout()
        plt.show()
    
    def plot_roc_curves(self, figsize=(10, 8)):
        """Построение ROC-кривых для каждого фолда"""
        plt.figure(figsize=figsize)
        
        for i, fold_result in enumerate(self.fold_predictions_):
            y_true = fold_result['y_true']
            y_proba = fold_result['y_proba']
            
            if y_proba is not None and len(np.unique(y_true)) > 1:
                fpr, tpr, _ = roc_curve(y_true, y_proba)
                auc_score = roc_auc_score(y_true, y_proba)
                plt.plot(fpr, tpr, label=f'Fold {i+1} (AUC = {auc_score:.3f})')
        
        plt.plot([0, 1], [0, 1], 'k--', label='Random classifier')
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title('ROC Curves by Fold')
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.show()
    
    def get_best_model(self, metric='roc_auc'):
        """Возвращает модель с лучшими показателями по заданной метрике"""
        if metric not in self.cv_results_:
            raise ValueError(f"Метрика {metric} не найдена")
        
        best_fold = np.argmax(self.cv_results_[metric])
        return self.models_[best_fold]
    
    def get_predictions(self):
        """Возвращает объединенные предсказания по всем фолдам"""
        y_true_full = np.concatenate([fold['y_true'] for fold in self.fold_predictions_])
        y_pred_full = np.concatenate([fold['y_pred'] for fold in self.fold_predictions_])
        y_proba_full = np.concatenate([fold['y_proba'] for fold in self.fold_predictions_ 
                                     if fold['y_proba'] is not None])
        
        return y_true_full, y_pred_full, y_proba_full
```

```python
import numpy as np
import pandas as pd
import torch
from torch.utils.data import Dataset, DataLoader
from sklearn.utils import check_random_state
class FluorescenceSpectraDataset(Dataset):
    def __init__(self, data, target_column=None, augmentation=False, 
                 augmentation_params=None, random_state=None):
        """
        Custom Dataset for fluorescence spectra with built-in augmentation.
        
        Args:
            data: Input data (numpy array or pandas DataFrame)
            target_column: Name of target column (if data is DataFrame)
            augmentation: Whether to apply augmentation
            augmentation_params: Parameters for augmentation
                - augment_prob: probability of applying augmentation (default: 0.5)
                - noise_scale: scale of Gaussian noise (default: 0.01)
                - stretch_range: range for random stretching (default: (0.95, 1.05))
                - shift_range: range for random shifting (default: (-0.05, 0.05))
            random_state: Random seed for reproducibility
        """
        self.augmentation = augmentation
        self.rng = check_random_state(random_state)
        
        if isinstance(data, pd.DataFrame):
            if target_column is None:
                raise ValueError("target_column must be specified for DataFrame input")
            self.X = data.drop(columns=[target_column]).values
            self.y = data[target_column].values
            self.columns = data.columns.tolist()
        else:
            self.X = data
            self.y = None
            self.columns = None
        
        # Set default augmentation parameters
        self.aug_params = {
            'augment_prob': 0.5,
            'noise_scale': 0.01,
            'stretch_range': (0.95, 1.05),
            'shift_range': (-0.05, 0.05),
            **({} if augmentation_params is None else augmentation_params)
        }
        
    def __len__(self):
        return len(self.X)
    
    def __getitem__(self, idx):
        spectrum = self.X[idx].astype(np.float32)
        target = self.y[idx] if self.y is not None else -1
        
        if self.augmentation and self.rng.rand() < self.aug_params['augment_prob']:
            # Apply random stretching
            stretch_factor = self.rng.uniform(*self.aug_params['stretch_range'])
            x_original = np.arange(len(spectrum))
            x_new = np.linspace(0, len(spectrum), int(len(spectrum) * stretch_factor))
            spectrum = np.interp(x_new, x_original, spectrum)
            
            # Apply random shifting
            shift = self.rng.uniform(*self.aug_params['shift_range'])
            spectrum = spectrum * (1 + shift)
            
            # Add Gaussian noise
            noise = self.rng.normal(0, self.aug_params['noise_scale'], len(spectrum))
            spectrum = spectrum + noise
            
            # Normalize to original range if needed
            spectrum = np.clip(spectrum, 0, None)  # assuming non-negative spectra
            
            # Resample back to original length if stretching changed it
            if len(spectrum) != len(self.X[idx]):
                spectrum = np.interp(
                    np.linspace(0, len(spectrum)-1, len(self.X[idx])),
                    np.arange(len(spectrum)),
                    spectrum
                )
        
        # Convert to torch tensors
        spectrum_tensor = torch.FloatTensor(spectrum)
        target_tensor = torch.LongTensor([target]) if self.y is not None else torch.Tensor([-1])
        
        return spectrum_tensor, target_tensor


class FluorescenceDataLoader(DataLoader):
    def __init__(self, data, target_column=None, batch_size=32, shuffle=True, 
                 augmentation=False, augmentation_params=None, random_state=None, **kwargs):
        """
        Custom DataLoader for fluorescence spectra with built-in augmentation.
        
        Args:
            data: Input data (numpy array or pandas DataFrame)
            target_column: Name of target column (if data is DataFrame)
            batch_size: Number of samples per batch
            shuffle: Whether to shuffle the data
            augmentation: Whether to apply augmentation
            augmentation_params: Parameters for augmentation
            random_state: Random seed for reproducibility
            **kwargs: Additional arguments for DataLoader
        """
        dataset = FluorescenceSpectraDataset(
            data=data,
            target_column=target_column,
            augmentation=augmentation,
            augmentation_params=augmentation_params,
            random_state=random_state
        )
        
        super().__init__(
            dataset,
            batch_size=batch_size,
            shuffle=shuffle,
            **kwargs
        )
```

```python
import random
import torch.nn as nn
from torch.utils.data import TensorDataset
class SpectralClassifierFCNN(nn.Module):
    def __init__(self, 
                 in_dim=147, 
                 hidden_dims=[256, 128, 64], 
                 out_dim=1, 
                 activation="leaky_relu", 
                 dropout_rate=0.3, 
                 random_seed=None):
        super().__init__()
        
        activations = {
            "leaky_relu": nn.LeakyReLU(0.2),
            "relu": nn.ReLU(),
            "selu": nn.SELU(),
            "gelu": nn.GELU(),
            "sigmoid": nn.Sigmoid(),
            "silu": nn.SiLU()
        }
        if random_seed is not None:
            self.set_seed(random_seed) 
        
        self.first_block = nn.Sequential(
            nn.Linear(in_dim, hidden_dims[0]),
            nn.LayerNorm(hidden_dims[0]),
            nn.Dropout(dropout_rate),
            activations[activation]
        )
        
        self.hidden_blocks = nn.ModuleList()
        for i in range(len(hidden_dims)-1):
            block = nn.Sequential(
                nn.Linear(hidden_dims[i], hidden_dims[i+1]),
                nn.LayerNorm(hidden_dims[i+1]),
                nn.Dropout(dropout_rate),
                activations[activation]
            )
            self.hidden_blocks.append(block)
        
        self.output = nn.Sequential(
            nn.Linear(hidden_dims[-1], out_dim),
        )

        self._init_weights()

    def set_seed(self, seed=42):
        torch.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
        np.random.seed(seed)
        random.seed(seed)
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False
    
    def predict(self, val_dataloader):
        if isinstance(val_dataloader, DataLoader):
            pass
        elif isinstance(val_dataloader, TensorDataset):
            val_dataloader = DataLoader(val_dataloader, 
                                        batch_size=64, 
                                        shuffle=False)
                
        device = "cuda" \
            if next(self.parameters()).is_cuda \
            else "cpu"
        self.to(device)
        self.eval()
        preds_all = np.array([])
        proba_all = np.array([])
        labels_all = np.array([])
        for batch, labels in val_dataloader:
            with torch.no_grad():
                batch = batch.to(torch.float)
                logits = self(batch.to(device)).squeeze().to(torch.float)
                proba = torch.sigmoid(logits)
                preds = (proba > 0.5).to(torch.int)

                preds_all = np.concatenate((preds_all, 
                                            preds.detach().cpu().numpy()))
                labels_all = np.concatenate((labels_all, 
                                             labels.detach().cpu().numpy()))
                proba_all = np.concatenate((proba_all, 
                                            proba.detach().cpu().numpy()))

        return preds_all, proba_all, labels_all
    def create_default_dataloader(self, 
                                  X: pd.DataFrame, 
                                  y: pd.Series, 
                                  batch_size: int=32, 
                                  shuffle: bool=True, 
                                  worker_seed=42) -> DataLoader:

        labels = torch.tensor(y.to_numpy())
        data = torch.tensor(X.to_numpy(), dtype=torch.float)

        dataset = TensorDataset(data, labels)
        dataloader = DataLoader(dataset, 
                                batch_size=batch_size, 
                                shuffle=shuffle, 
                                worker_init_fn=worker_seed)
        return dataloader
    
    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Linear):
                nn.init.kaiming_normal_(m.weight, 
                                        mode='fan_out', 
                                        nonlinearity='leaky_relu')
                nn.init.constant_(m.bias, 0)
    
    def forward(self, x):
        x = self.first_block(x)
        for block in self.hidden_blocks:
            residual = x
            x = block(x)
            if x.shape == residual.shape:
                x = x + residual
        return self.output(x)
```

```python
from tqdm import tqdm
from sklearn.metrics import accuracy_score, confusion_matrix
def train(model, epochs, 
          optim, criterion, 
          train_dataloader, val_dataloader, 
          scheduler=None, logging=False, 
          tqdm_desc="Processed") -> tuple:
    losses_train = []
    losses_val = []
    acc = []
    precision = []
    recall = []
    specificity = []
    device = "cuda" if next(model.parameters()).is_cuda else "cpu"
    for epoch in tqdm(range(epochs), 
                      desc=tqdm_desc):
        loss_train = []
        loss_val = []
        model.train()
        for batch, labels in train_dataloader:
            labels = labels.squeeze()
            optim.zero_grad()
            outputs = model(batch)
            loss = criterion(outputs.to(torch.float).squeeze(), 
                             labels.to(torch.float))
            loss.backward()
            optim.step()

            loss_train.append(loss.detach().cpu().numpy())

        losses_train.append(
            sum(loss_train) / len(loss_train)
            )

        model.eval()
        preds_epoch = np.array([])
        labels_epoch = np.array([])
        for batch, labels in val_dataloader:
            labels = labels.squeeze()
            with torch.no_grad():
                batch = batch.to(torch.float)
                logits = model(
                    batch.to(device)
                    ).squeeze().to(torch.float)
                preds = (torch.sigmoid(logits) > 0.5).float()
                loss = criterion(logits, 
                                 labels.to(torch.float).to(device)
                                 )
                loss_val.append(loss.detach().cpu().numpy())
                preds_epoch = np.concatenate(
                    (preds_epoch, preds.detach().cpu().numpy())
                    )
                labels_epoch = np.concatenate(
                    (labels_epoch, labels.detach().cpu().numpy())
                    )

        tn, fp, fn, tp = confusion_matrix(labels_epoch, preds_epoch).ravel()
        losses_val.append(sum(loss_val) / len(loss_val))
        acc.append(accuracy_score(labels_epoch, preds_epoch))
        specificity.append(tn / (tn + fp))
        precision.append(tp / (tp + fp))
        recall.append(tp / (tp + fn))
        if scheduler != None:
            if str(scheduler.__class__) == \
                "<class 'torch.optim.lr_scheduler.ReduceLROnPlateau'>":
                scheduler.step(epoch)
            else:
                scheduler.step()


    print(f"Training complete successfully. Best accuracy on validation: {max(acc):.4f}")
    return losses_train, losses_val, acc, precision, recall, specificity
```

```python
import torch.nn.functional as F
class FocalLoss(nn.Module):
    def __init__(self, alpha=0.8, gamma=2.0):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma

    def forward(self, inputs, targets):
        BCE_loss = F.binary_cross_entropy_with_logits(inputs, 
                                                      targets, 
                                                      reduction='none')
        pt = torch.exp(-BCE_loss)
        F_loss = self.alpha * (1-pt)**self.gamma * BCE_loss
        return F_loss.mean()
```
